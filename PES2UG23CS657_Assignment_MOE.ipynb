```python
import os
from getpass import getpass

os.environ["GROQ_API_KEY"] = getpass("Enter your Groq API Key: ")
```

```python
from groq import Groq

client = Groq(api_key=os.environ["GROQ_API_KEY"])

MODEL_NAME = "llama-3.1-8b-instant"
```

```python
MODEL_CONFIG = {
    "technical": {
        "system_prompt": """You are a Technical Support Expert.
Be precise, code-focused, and rigorous.
Provide corrected code snippets when needed."""
    },
    "billing": {
        "system_prompt": """You are a Billing Support Expert.
Be empathetic and professional.
Clearly explain refund policies and next steps."""
    },
    "general": {
        "system_prompt": """You are a helpful general customer support assistant."""
    },
    "tool": {
        "system_prompt": "You are a data retrieval system."
    }
}
```

```python
def route_prompt(user_input: str) -> str:
    routing_prompt = f"""
Classify the following text into one of these categories:
[technical, billing, general, tool]

If it asks for real-time data like prices, weather, or current info, classify as: tool.

Return ONLY the category name.

Text:
{user_input}
"""

    response = client.chat.completions.create(
        model=MODEL_NAME,
        temperature=0,
        messages=[
            {"role": "system", "content": "You are a strict classifier."},
            {"role": "user", "content": routing_prompt}
        ]
    )

    return response.choices[0].message.content.strip().lower()
```

```python
def get_bitcoin_price():
    return "Bitcoin is currently around $60,000 (mock data)."
```

```python
def process_request(user_input: str) -> str:
    category = route_prompt(user_input)
    print("Router Decision:", category)

    if category == "tool":
        if "bitcoin" in user_input.lower():
            return get_bitcoin_price()
        else:
            return "Tool requested but no matching function available."

    system_prompt = MODEL_CONFIG.get(category, MODEL_CONFIG["general"])["system_prompt"]

    response = client.chat.completions.create(
        model=MODEL_NAME,
        temperature=0.7,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_input}
        ]
    )

    return response.choices[0].message.content
```

```python
print(process_request("My python script throws an IndexError."))
print(process_request("I was charged twice this month."))
print(process_request("What are your business hours?"))
print(process_request("What is the current price of Bitcoin?"))
```
